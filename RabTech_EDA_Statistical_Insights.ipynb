```python

# ============================================================
# RABTECH ACADEMY - TASK 04
# Exploratory Data Analysis & Statistical Insights
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown
from scipy import stats

warnings.filterwarnings("ignore")

# 1. LOAD DATASET
possible_files = [
    "/content/clean_dataset.csv",
    "/content/retail-orders-raw.csv",
    "clean_dataset.csv",
    "retail-orders-raw.csv"
]

file_path = next((f for f in possible_files if os.path.exists(f)), None)

if file_path is None:
    raise FileNotFoundError(
        "Dataset not found. Upload retail-orders-raw.csv "
        "or clean_dataset.csv to Google Colab first."
    )

df = pd.read_csv(file_path)
df.columns = df.columns.str.strip().str.lower()

print("=" * 65)
print("RABTECH TASK 04: EXPLORATORY DATA ANALYSIS")
print("=" * 65)
print("Dataset loaded:", file_path)
print("Original shape:", df.shape)
print("\nFirst five records:")
display(df.head())

# 2. CLEAN DATA TYPES FOR ANALYSIS
if "order_date" in df.columns:
    # Handle ISO dates and day/month/year dates separately
    dates = df["order_date"].astype("string").str.strip()
    parsed = pd.to_datetime(dates, format="%Y-%m-%d", errors="coerce")
    other = parsed.isna()
    parsed.loc[other] = pd.to_datetime(
        dates.loc[other], format="%d/%m/%Y", errors="coerce"
    )
    # Handle any remaining common date formats
    other = parsed.isna()
    parsed.loc[other] = pd.to_datetime(
        dates.loc[other], errors="coerce"
    )
    df["order_date"] = parsed

numeric_candidates = [
    "quantity", "unit_price", "discount_pct",
    "gross_sales", "discount_amount", "net_sales",
    "profit", "profit_margin"
]

for col in numeric_candidates:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Create sales features if they are not already available
if {"quantity", "unit_price"}.issubset(df.columns):
    if "gross_sales" not in df.columns:
        df["gross_sales"] = df["quantity"] * df["unit_price"]

    if "discount_pct" in df.columns:
        df["discount_pct"] = df["discount_pct"].where(
            df["discount_pct"].between(0, 100)
        ).fillna(0)
        if "discount_amount" not in df.columns:
            df["discount_amount"] = (
                df["gross_sales"] * df["discount_pct"] / 100
            )
    else:
        df["discount_amount"] = 0

    if "net_sales" not in df.columns:
        df["net_sales"] = df["gross_sales"] - df["discount_amount"]

# Keep a numeric version of usable data
numeric_df = df.select_dtypes(include=np.number)
if numeric_df.empty:
    raise ValueError("No numeric columns available for statistical analysis.")

# 3. SUMMARY STATISTICS
print("\n" + "=" * 65)
print("1. SUMMARY STATISTICS")
print("=" * 65)

summary = numeric_df.describe(
    percentiles=[0.25, 0.50, 0.75]
).T
summary["median"] = numeric_df.median()
summary["missing_values"] = numeric_df.isna().sum()
summary["unique_values"] = numeric_df.nunique()

display(summary.round(2))
summary.to_csv("/content/task4_summary_statistics.csv")

print("\nMissing values by column:")
display(df.isna().sum().rename("Missing Values").to_frame())

# 4. CORRELATION MATRIX AND HEATMAP
print("\n" + "=" * 65)
print("2. CORRELATION ANALYSIS")
print("=" * 65)

corr = numeric_df.corr(method="pearson")
display(corr.round(2))

if corr.shape[1] >= 2:
    fig, ax = plt.subplots(figsize=(max(7, len(corr.columns)), 5))
    image = ax.imshow(corr, vmin=-1, vmax=1, cmap="coolwarm")
    ax.set_xticks(range(len(corr.columns)))
    ax.set_yticks(range(len(corr.columns)))
    ax.set_xticklabels(corr.columns, rotation=45, ha="right")
    ax.set_yticklabels(corr.columns)
    for i in range(len(corr.columns)):
        for j in range(len(corr.columns)):
            value = corr.iloc[i, j]
            if pd.notna(value):
                ax.text(j, i, f"{value:.2f}", ha="center", va="center",
                        fontsize=8)
    ax.set_title("Correlation Heatmap")
    fig.colorbar(image, ax=ax, label="Pearson correlation")
    plt.tight_layout()
    plt.savefig("/content/task4_correlation_heatmap.png", dpi=150)
    plt.show()
else:
    print("Heatmap skipped: at least two numeric columns are required.")

# 5. HISTOGRAMS
print("\n" + "=" * 65)
print("3. HISTOGRAM DISTRIBUTIONS")
print("=" * 65)

for col in numeric_df.columns:
    values = numeric_df[col].dropna()
    if len(values) == 0:
        continue
    plt.figure(figsize=(7, 4))
    plt.hist(values, bins=min(10, max(1, values.nunique())), edgecolor="black")
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    safe_col = "".join(c if c.isalnum() else "_" for c in col)
    plt.savefig(f"/content/task4_histogram_{safe_col}.png", dpi=120)
    plt.show()

# 6. BOXPLOTS AND IQR OUTLIER CHECK
print("\n" + "=" * 65)
print("4. BOXPLOTS AND OUTLIER DETECTION")
print("=" * 65)

outlier_results = []

for col in numeric_df.columns:
    values = numeric_df[col].dropna()
    if len(values) == 0:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = int(((values < lower) | (values > upper)).sum())

    outlier_results.append({
        "column": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "lower_bound": lower,
        "upper_bound": upper,
        "potential_outliers": count
    })

    plt.figure(figsize=(7, 3))
    plt.boxplot(values, vert=False)
    plt.title(f"Box Plot of {col}")
    plt.xlabel(col)
    plt.tight_layout()
    safe_col = "".join(c if c.isalnum() else "_" for c in col)
    plt.savefig(f"/content/task4_boxplot_{safe_col}.png", dpi=120)
    plt.show()

outlier_table = pd.DataFrame(outlier_results)
display(outlier_table.round(2))
outlier_table.to_csv("/content/task4_outlier_check.csv", index=False)

print("Note: IQR flags potential outliers; it does not automatically remove them.")

# 7. THREE BUSINESS HYPOTHESES
print("\n" + "=" * 65)
print("5. BUSINESS HYPOTHESIS TESTING")
print("=" * 65)

hypothesis_results = []

def add_result(name, h0, test_name, statistic, p_value, n, note=""):
    if pd.notna(p_value):
        conclusion = (
            "Reject H0 at 5% significance"
            if p_value < 0.05
            else "Fail to reject H0 at 5% significance"
        )
    else:
        conclusion = "Insufficient data to test"
    hypothesis_results.append({
        "hypothesis": name,
        "null_hypothesis": h0,
        "test": test_name,
        "statistic": statistic,
        "p_value": p_value,
        "sample_size": n,
        "conclusion": conclusion,
        "notes": note
    })

# Hypothesis 1: Discount percentage and net sales are associated
if {"discount_pct", "net_sales"}.issubset(df.columns):
    temp = df[["discount_pct", "net_sales"]].dropna()
    if len(temp) >= 3 and temp["discount_pct"].nunique() > 1 and temp["net_sales"].nunique() > 1:
        r, p = stats.spearmanr(temp["discount_pct"], temp["net_sales"])
        add_result(
            "Discount percentage is associated with net sales",
            "There is no monotonic association between discount percentage and net sales",
            "Spearman rank correlation", r, p, len(temp),
            "Association does not establish causation."
        )
    else:
        add_result(
            "Discount percentage is associated with net sales",
            "There is no monotonic association between discount percentage and net sales",
            "Spearman rank correlation", np.nan, np.nan, len(temp),
            "Insufficient variation or fewer than 3 complete records."
        )
else:
    add_result(
        "Discount percentage is associated with net sales",
        "There is no monotonic association between discount percentage and net sales",
        "Spearman rank correlation", np.nan, np.nan, 0,
        "Required columns are missing."
    )

# Hypothesis 2: Net sales differ across payment statuses
if {"payment_status", "net_sales"}.issubset(df.columns):
    temp = df[["payment_status", "net_sales"]].dropna()
    groups = [
        g["net_sales"].values
        for _, g in temp.groupby("payment_status")
        if len(g) >= 2
    ]
    if len(groups) >= 2:
        statistic, p = stats.kruskal(*groups)
        add_result(
            "Net sales differ by payment status",
            "Net sales distributions are the same across payment statuses",
            "Kruskal-Wallis test", statistic, p, len(temp),
            "Only groups with at least 2 records are tested."
        )
    else:
        add_result(
            "Net sales differ by payment status",
            "Net sales distributions are the same across payment statuses",
            "Kruskal-Wallis test", np.nan, np.nan, len(temp),
            "Need at least 2 payment-status groups with 2 or more records each."
        )
else:
    add_result(
        "Net sales differ by payment status",
        "Net sales distributions are the same across payment statuses",
        "Kruskal-Wallis test", np.nan, np.nan, 0,
        "Required columns are missing."
    )

# Hypothesis 3: Order quantity and unit price are associated
if {"quantity", "unit_price"}.issubset(df.columns):
    temp = df[["quantity", "unit_price"]].dropna()
    if len(temp) >= 3 and temp["quantity"].nunique() > 1 and temp["unit_price"].nunique() > 1:
        r, p = stats.spearmanr(temp["quantity"], temp["unit_price"])
        add_result(
            "Order quantity is associated with unit price",
            "There is no monotonic association between quantity and unit price",
            "Spearman rank correlation", r, p, len(temp),
            "This tests association, not a causal relationship."
        )
    else:
        add_result(
            "Order quantity is associated with unit price",
            "There is no monotonic association between quantity and unit price",
            "Spearman rank correlation", np.nan, np.nan, len(temp),
            "Insufficient variation or fewer than 3 complete records."
        )
else:
    add_result(
        "Order quantity is associated with unit price",
        "There is no monotonic association between quantity and unit price",
        "Spearman rank correlation", np.nan, np.nan, 0,
        "Required columns are missing."
    )

hypotheses = pd.DataFrame(hypothesis_results)
display(hypotheses.round(4))
hypotheses.to_csv("/content/task4_hypothesis_tests.csv", index=False)

# 8. TOP FIVE DATA-DRIVEN FINDINGS
print("\n" + "=" * 65)
print("6. TOP FIVE FINDINGS")
print("=" * 65)

findings = []

findings.append(
    f"The analyzed dataset contains {len(df)} records and {len(df.columns)} columns."
)

if "net_sales" in df.columns and df["net_sales"].notna().any():
    sales = df["net_sales"].dropna()
    findings.append(
        f"Net sales: mean = {sales.mean():.2f}, median = {sales.median():.2f}, "
        f"minimum = {sales.min():.2f}, maximum = {sales.max():.2f}."
    )

if "payment_status" in df.columns:
    counts = df["payment_status"].astype("string").value_counts()
    if len(counts):
        findings.append(
            f"Most common payment status: {counts.index[0]} ({counts.iloc[0]} records)."
        )

if "category" in df.columns:
    counts = df["category"].astype("string").value_counts()
    if len(counts):
        findings.append(
            f"Most frequent category: {counts.index[0]} ({counts.iloc[0]} records)."
        )

if "city" in df.columns:
    counts = df["city"].astype("string").value_counts()
    if len(counts):
        findings.append(
            f"Most frequent city: {counts.index[0]} ({counts.iloc[0]} records)."
        )

if len(findings) < 5 and not corr.empty and corr.shape[1] >= 2:
    pairs = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool)).stack().dropna()
    if len(pairs):
        pair = pairs.abs().idxmax()
        findings.append(
            f"Strongest absolute numeric correlation: {pair[0]} and {pair[1]} "
            f"(r = {corr.loc[pair[0], pair[1]]:.2f}). Correlation is not causation."
        )

if len(findings) < 5 and not outlier_table.empty:
    total_outliers = int(outlier_table["potential_outliers"].sum())
    findings.append(
        f"IQR analysis flagged {total_outliers} potential outlier value(s) "
        "across numeric columns; inspect box plots before deciding what to do."
    )

if len(findings) < 5:
    findings.append(
        "The dataset is small, so findings and statistical tests should be treated as preliminary."
    )

findings = findings[:5]
for i, finding in enumerate(findings, 1):
    print(f"{i}. {finding}")

# 9. EXPORT REPORTS AND FINDINGS
with open("/content/task4_top5_findings.md", "w", encoding="utf-8") as f:
    f.write("# Task 04: Top Five EDA Findings\n\n")
    for i, finding in enumerate(findings, 1):
        f.write(f"{i}. {finding}\n\n")
    f.write(
        "\n## Limitations\n"
        "- This analysis uses only the records supplied by RabTech Academy.\n"
        "- A small dataset may not provide enough statistical power for hypothesis testing.\n"
        "- Correlation does not prove causation.\n"
        "- IQR identifies potential outliers; business context is needed before removal.\n"
    )

print("\n" + "=" * 65)
print("TASK 04 ANALYSIS FINISHED")
print("=" * 65)
print("Files created in /content:")
for name in sorted(os.listdir("/content")):
    if name.startswith("task4_"):
        print(" -", name)

print(
    "\nSubmission tip: keep this notebook with the charts and outputs, "
    "then upload the .ipynb file and the report CSV/Markdown files to your public GitHub repository."
)
```
